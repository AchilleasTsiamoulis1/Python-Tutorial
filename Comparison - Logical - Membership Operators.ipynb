# Comparison Operators

```python
10 != 50
```

```python
10 == 10
```

```python
'Vanilla' != 'Chocolate'
```

```python
x = 'Vanilla'
y = 'Chocolate'

x == y
```

```python
10 < 50
```

```python
10 <= 10
```

```python
10 >= 50
```

```python
'Vanilla' > 'Chocolate'
```

# Logical Operators

```python
(10 > 50) and (50 > 10)
```

```python
(10 > 50) or (50 > 10)
```

```python
('Vanilla' > 'Chocolate') or (50 > 10)
```

```python
not(50 > 10)
```

# Membership Operatos

```python
ice_cream = 'I love Chocolate ice cream'

'love' in ice_cream
```

```python
'love' in 'I love Chocolate ice cream'
```

```python
scoops = [1,2,3,4,5]

2 in scoops
```

```python
scoops = [1,2,3,4,5]

6 in scoops
```

```python
scoops = [1,2,3,4,5]
wanted_scoops = 8

wanted_scoops in scoops
```
