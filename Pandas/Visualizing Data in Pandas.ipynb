# Pandas Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Achilleas\Desktop\24_Hour_Course\Python\Pandas\Ice Cream Ratings.csv")
df = df.set_index('Date')
df
```

```python
print(plt.style.available)
plt.style.use('ggplot')
```

```python
df.plot(kind = 'line', title = 'Ice Cream Ratings', xlabel = 'Daily Rating', ylabel = 'Scores')
```

```python
df.plot(kind = 'bar', stacked = True)
```

```python
# Vertical Bar
df.plot.barh(stacked = True)
```

```python
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating', s = 350, c = 'Cyan')
```

```python
df.plot.hist(bins = 15)
```

```python
df.boxplot()
```

```python
df.plot.area(figsize = (10, 5))
```

```python
df.plot.pie(y = 'Flavor Rating', figsize = (10, 6))
```
